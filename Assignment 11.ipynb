```python
import pandas as pd
import numpy as np
df = pd.read_csv("cars.csv")
```

#### Check Duplicate Records

```python
df[df.duplicated()]
```

#### Drop Duplicate Records

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

#### Check missing or null values

```python
df.isnull().sum()
```

#### Drop missing or null values

```python
df.dropna(inplace=True)
df.isnull().sum()
```

#### Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
df.groupby('Make')['MSRP'].mean().head(10).round().reset_index()
```

#### Show the average MSRP (price) per years (rounded figure)?

```python
df.groupby('Year')['MSRP'].mean().round().reset_index()
```

#### Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
df['MPG'] = (df['city mpg'] + df['highway MPG']) / 2

df.groupby('Year')[['Engine HP', 'MPG']].mean().round()
```

#### What percentage of cars are Compact, Midsize, or Large?

```python
df['Vehicle Size'].value_counts(normalize=True).mul(100).round(2).loc[['Compact', 'Midsize', 'Large']].reset_index()
```

#### Which vehicle styles are the most sold (most common) in the dataset?

```python
df['Vehicle Style'].value_counts().reset_index()
```

#### Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
df[(df['MSRP'] >= 40000) & (df['Popularity'] >= 1000)]['Make'].value_counts().reset_index()
```
