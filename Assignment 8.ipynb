#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
# Employee Class

class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Method to display employee details
    def describe_employee(self):
        print("\nEmployee Details")
        print("----------------------------")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    # Method to update experience
    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to {self.year_of_experience} years.")

    # Method to increase salary
    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s new salary is {self.salary}.")


# Create an instance employee101
employee101 = Employee(
    "Ali Khan",
    101,
    "Karachi",
    50000,
    "Software Engineer",
    2020,
    "BS Computer Science",
    5
)

# Print four attributes individually
print("Name:", employee101.name)
print("Number:", employee101.number)
print("Salary:", employee101.salary)
print("Post:", employee101.post)

# Call methods
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)


# Create 10 different employee instances
employees = [
    Employee("Ahmed", 102, "Lahore", 45000, "Accountant", 2019, "B.Com", 6),
    Employee("Sara", 103, "Islamabad", 60000, "Manager", 2018, "MBA", 8),
    Employee("Usman", 104, "Peshawar", 40000, "Clerk", 2021, "BA", 3),
    Employee("Ayesha", 105, "Quetta", 70000, "HR Manager", 2017, "MBA", 9),
    Employee("Bilal", 106, "Multan", 55000, "Developer", 2020, "BS IT", 5),
    Employee("Hina", 107, "Faisalabad", 48000, "Teacher", 2019, "M.Ed", 6),
    Employee("Zain", 108, "Hyderabad", 52000, "Designer", 2022, "BS Design", 2),
    Employee("Fatima", 109, "Sialkot", 65000, "Doctor", 2016, "MBBS", 10),
    Employee("Hamza", 110, "Rawalpindi", 58000, "Engineer", 2018, "BE Civil", 7),
    Employee("Noor", 111, "Gujranwala", 47000, "Nurse", 2021, "BS Nursing", 4)
]

# Display details and perform updates
for emp in employees:
    emp.describe_employee()
    emp.increase_salary(3000)
    emp.update_year_of_experience(1)
```
